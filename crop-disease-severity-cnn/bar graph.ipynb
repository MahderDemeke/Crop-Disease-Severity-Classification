```python
import os
dir="OneDrive/Documents/Organaized data For research/Research Data/TrainDataRowData1"
list = os.listdir(dir) # dir is your directory path
image_no=[]
catagory=[]
for path,i in enumerate (list):
    path1=os.listdir(dir+'/'+i)
    number_of_files=len(path1)
    print(i,number_of_files)
    image_no.append(number_of_files)
    catagory.append(i)
print('image_no:',image_no)
print('class name:',catagory)
```

```python
#data.pivot_table(index='buy', afunc='size').plot(kind='bar')
import numpy as np
import matplotlib.pyplot as plt
ax = plt.figure(figsize = (800, 100))
fig = plt.figure()
X=np.arange(start=0, stop=20, step=1)
ax = fig.add_axes([0,0,1,1])
ax.bar(X,image_no, width = 0.5,color ='g')#cataory
ax.set_xticks(X,catagory)
for tick in ax.get_xticklabels():
    tick.set_rotation(90)
plt.savefig('OneDrive/Desktop/diagram folder/OriginalDataDistribution.png',dpi=150,bbox_inches='tight')
plt.show()
```

```python
#Py chart
from matplotlib import pyplot as plt
import numpy as np
fig = plt.figure()
ax = fig.add_axes([0,0,1,1])
ax.axis('equal')
NUM_COLORS = len(catagory)
ax.pie(image_no,autopct='%10.2f%%')
plt.show()
```

```python
import matplotlib.pyplot as plt

fig1, ax1 = plt.subplots(figsize=(7,6))
fig1.subplots_adjust(0.7,0,1.5,1)
theme = plt.get_cmap('bwr')
ax1.set_prop_cycle("color", [theme(1. * i / len(catagory)) for i in range(len(catagory))])

_, _ = ax1.pie(image_no, startangle=90)

ax1.axis('equal')

total = sum(image_no)
print(total)
plt.legend(
    loc='upper left',
    labels=['%s, %2.6f%%' % (
        l, (float(s) / total) * 100) for l, s in zip(catagory, image_no)],
    prop={'size': 14},
    bbox_to_anchor=(0.0, 1),
    bbox_transform=fig1.transFigure
)
plt.savefig('OneDrive/Desktop/diagram folder/OriginalDataDistributionpie.png',dpi=150,bbox_inches='tight')
plt.show()
```

```python
import matplotlib.pyplot as plt

sizes = [15, 30, 45, 10, 10, 24, 13, 18, 28, 20, 13, 15, 5, 1, 18, 10,
         10, 10]
labels = ["Frogs %s" % i for i in sizes]

fig1, ax1 = plt.subplots(figsize=(7, 5))
fig1.subplots_adjust(0.3,0,1,1)


theme = plt.get_cmap('bwr')
ax1.set_prop_cycle("color", [theme(1. * i / len(sizes)) for i in range(len(sizes))])

_, _ = ax1.pie(sizes, startangle=90)

ax1.axis('equal')

total = sum(sizes)
plt.legend(
    loc='upper left',
    labels=['%s, %10.1f%%' % (
        l, (float(s) / total) * 100) for l, s in zip(labels, sizes)],
    prop={'size': 110},
    bbox_to_anchor=(0.0, 1),
    bbox_transform=fig1.transFigure
)

plt.show()
```
